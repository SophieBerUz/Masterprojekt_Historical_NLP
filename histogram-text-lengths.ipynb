```python
import pandas as pd
# read dataset
df = pd.read_json('500_samples_perBucket_data.json', lines=True)
df['text'].head()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# calculate length of each text in column 'text'
df['text_length'] = df['text'].apply(len)
#print(df.iloc[162])
print(df.iloc[10])
```

```python
#extra code, clear input data for aya model

# Count the number of texts where the length exceeds 8319 characters
count = (df['text_length'] > 8319).sum()

print(f"Number of texts longer than 8319 characters: {count}")

df_filtered = df[df['text_length'] <= 8319]

# Optionally, drop the 'text_length' column if you no longer need it
df_filtered = df_filtered.drop(columns=['text_length'])

# store as json
df_filtered.to_json('500_samples_DATA_removedTextsLarger8319characters.json', orient='records', lines=True)
```

```python
#  Subplots for different areas of the text lengths
fig, axs = plt.subplots(2, 2, figsize=(12,10))  # 2x2 Subplots

# Split 1: Texts with length 0-250
axs[0, 0].hist(df[df['text_length'] <= 250]['text_length'], bins=20, edgecolor='black')
axs[0, 0].set_title('text length: 0-250 characters')
axs[0, 0].set_xlabel('number of characters in string')
axs[0, 0].set_ylabel('number of texts')

# Split 2: Texts with length 250-500
axs[0, 1].hist(df[(df['text_length'] > 250) & (df['text_length'] <= 500)]['text_length'], bins=20, edgecolor='black')
axs[0, 1].set_title('text length: 250-500 characters')
axs[0, 1].set_xlabel('number of characters in string')
axs[0, 1].set_ylabel('number of texts')

# Split 3: Texts with length 500-1000
axs[1, 0].hist(df[(df['text_length'] > 500) & (df['text_length'] <= 1000)]['text_length'], bins=20, edgecolor='black')
axs[1, 0].set_title('text length: 500-1000 characters')
axs[1, 0].set_xlabel('number of characters in string')
axs[1, 0].set_ylabel('number of texts')

# Split 4: Texts with length >1000
axs[1, 1].hist(df[df['text_length'] > 1000]['text_length'], bins=20, edgecolor='black')
axs[1, 1].set_title('text length: >1000 characters')
axs[1, 1].set_xlabel('number of characters in string')
axs[1, 1].set_ylabel('number of texts')

# add line to histogram for length 8319->could be handled by the model
axs[1, 1].axvline(x=8319, color='red', linestyle='--', linewidth=2, label='Länge 8319')
axs[1, 1].legend() 

# show subplots
plt.tight_layout()
plt.savefig('text_length_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import nltk
from nltk.tokenize import word_tokenize
# calculate number of tokens in each text
df['token_count'] = df['text'].apply(lambda x: len(word_tokenize(x)))
print(df.iloc[10])
# create subplots for each area of number of tokens  
fig, axs = plt.subplots(2, 2, figsize=(12, 10))  # 2x2 Subplots

# Split 1: Texts with number of tokens between 0-25
axs[0, 0].hist(df[df['token_count'] <= 25]['token_count'], bins=20, edgecolor='black')
axs[0, 0].set_title('number of tokens: 0-25 Tokens')
axs[0, 0].set_xlabel('number of tokens')
axs[0, 0].set_ylabel('number of texts')

# Split 2: Texts with number of tokens between 25-50
axs[0, 1].hist(df[(df['token_count'] > 25) & (df['token_count'] <= 50)]['token_count'], bins=20, edgecolor='black')
axs[0, 1].set_title('number of tokens: 25-50 Tokens')
axs[0, 1].set_xlabel('number of tokens')
axs[0, 1].set_ylabel('number of texts')

# Split 3: Texts with number of tokens between 50-100
axs[1, 0].hist(df[(df['token_count'] > 50) & (df['token_count'] <= 100)]['token_count'], bins=20, edgecolor='black')
axs[1, 0].set_title('number of tokens: 50-100 Tokens')
axs[1, 0].set_xlabel('number of tokens')
axs[1, 0].set_ylabel('number of texts')

# Split 4: Texts with number of tokens >100
axs[1, 1].hist(df[df['token_count'] > 100]['token_count'], bins=20, edgecolor='black')
axs[1, 1].set_title('number of tokens: >100 Tokens')
axs[1, 1].set_xlabel('number of tokens')
axs[1, 1].set_ylabel('number of texts')

# add vertical line at number of tokens = 8319 in the fourth subplot
axs[1, 1].axvline(x=1450, color='red', linestyle='--', linewidth=2, label='number of tokens 1450')
axs[1, 1].legend()

# show Subplots
plt.tight_layout()
plt.savefig('token_count_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```
