```python
from transformers import pipeline
import torch
import pandas as pd
```

```python
# Load the JSON dataset
df_corpus_texts = pd.read_json('dta_corpus_texts.json', lines=True)
df_corpus_tokens = pd.read_json('dta_corpus_tokens.json', lines=True)
```

```python
# concat both datasets
df_merged = pd.concat([df_corpus_texts, df_corpus_tokens], axis=1)
# remove duplicate column
df_merged = df_merged.loc[:,~df_merged.columns.duplicated()]

# store as json file
df_merged.to_json('poems_complete.json', orient='records', lines=True)
print("Der DataFrame wurde erfolgreich als JSON gespeichert.")

# print head of concated dataframe
print(df_merged.head())
```

```python
# define buckets
buckets = {
    '1497': (1497, 1497),
    '1600-1649': (1600, 1649),
    '1650-1699': (1650, 1699),
    '1700-1749': (1700, 1749),
    '1750-1799': (1750, 1799),
    '1800-1849': (1800, 1849),
    '1850-1899': (1850, 1899),
    '1913': (1913, 1913)
}

# function to sort years into buckets
def get_bucket(year):
    for bucket, (start, end) in buckets.items():
        if start <= year <= end:
            return bucket
    return 'Nicht zugeordnet'

# new coloum for buckets
df_merged['bucket'] = df_merged['year'].apply(get_bucket)

# count poems per bucket
bucket_counts = df_merged['bucket'].value_counts()

# print results
print("Anzahl der Gedichte pro Bucket:")
print(bucket_counts)
#print(df_merged.head())

# store as json file
df_merged.to_json('poems_complete_with_buckets.json', orient='records', lines=True)
print("Der DataFrame wurde erfolgreich als JSON gespeichert.")
```
